## The 14-Session Core Curriculum (Plan A)

The core track is organized into 4 cohesive engineering modules (~21 hours total, designed for ~4.5 weeks at 3 sessions/week):

### Module 1: Ingestion & Vectorized Signals (NumPy)
* 🟢 **[Session 01: Modern Environment & Fast Execution](./session_01_environment_ipython.ipynb)**  
  *Setting up with `uv`, measuring the memory reality of Python lists vs. C-buffers, the Execute-Explore workflow, and live introspection.*
* 🟢 **[Session 02: Multi-Channel Signals & NumPy ndarrays](./session_02_numpy_arrays.ipynb)**  
  *Matrix dimensions `(samples x channels)`, cutting RAM by 50% with `float32`, and calibrating thousands of sensors simultaneously via Broadcasting.*
* ⏳ **[Session 03: Boolean Masking, Views vs Copies & Signal Filtering](./session_03_numpy_filtering.ipynb)**  
  *Isolating sensor glitches, the critical difference between views and copies in slicing, and conditional clamping with `np.where`.*

### Module 2: Tabular Telemetry & Columnar Storage (pandas)
* ⏳ **[Session 04: pandas Series & DataFrame Fundamentals](./session_04_pandas_basics.ipynb)**  
  *Representing single sensor streams and building arrays, automatic index alignment across unsynchronized clocks.*
* ⏳ **[Session 05: Precise Indexing & Element Mapping](./session_05_pandas_indexing.ipynb)**  
  *Subsetting sensor channels with `loc`/`iloc`, applying non-linear conversions with modern `df.map`.*
* ⏳ **[Session 06: Reading High-Volume Logs in Chunks & Parquet Storage](./session_06_data_loading_chunks.ipynb)**  
  *Streaming large CSV exports with `chunksize`, exporting to compressed Parquet with `to_parquet()`, and measuring disk footprint.*

### Module 3: Cleaning & Wrangling Sensor Streams
* ⏳ **[Session 07: Handling Missing Telemetry & Dropouts](./session_07_missing_telemetry.ipynb)**  
  *Detecting dropped network packets, forward-filling (`ffill`) dead sensor channels, and interpolation.*
* ⏳ **[Session 08: Outlier Detection & Signal Transformations](./session_08_outliers_transforms.ipynb)**  
  *Deduplicating timestamps, clipping physical sensor limits, and bucketing continuous values with `cut`/`qcut`.*
* ⏳ **[Session 09: Merging Telemetry with Building Metadata](./session_09_merging_metadata.ipynb)**  
  *Joining sensor time series with asset/room tables using relational merges (`merge`, `join`).*
* ⏳ **[Session 10: Pivoting & Reshaping Telemetry Grids](./session_10_pivoting_telemetry.ipynb)**  
  *Transforming narrow event streams into wide sensor matrices (`pivot`, `pivot_table`, `melt`).*

### Module 4: Deep Time Series & Capstone Analysis (3 Sessions)
* ⏳ **[Session 11: DatetimeIndex, Frequency & Slicing](./session_11_timeseries_indexing.ipynb)**  
  *Parsing sensor timestamps with `pd.to_datetime`, timezone handling, date range generation, slicing exact time windows.*
* ⏳ **[Session 12: Resampling & Frequency Conversion](./session_12_timeseries_resampling.ipynb)**  
  *Downsampling high-frequency raw telemetry (10s/1m) to 15-minute energy averages (`resample('15min')`), upsampling with interpolation.*
* ⏳ **[Session 13: Rolling Windows, Shift/Lag & Anomaly Detection](./session_13_timeseries_rolling_anomalies.ipynb)**  
  *Moving averages (`rolling(window=60).mean()`), lag features with `.shift()`, detecting sudden electrical spikes and sensor drift.*
* ⏳ **[Session 14: Capstone: End-to-End Building Energy Audit](./session_14_capstone_energy_audit.ipynb)**  
  *Full pipeline on Building Data Genome 2: raw telemetry ingestion ➔ cleaning ➔ resampling ➔ anomaly flags ➔ peak load summary report.*
